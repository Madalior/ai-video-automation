## Step 2: Setup Virtual Display
Creates a fake monitor so Chrome works properly.

```python
from pyvirtualdisplay import Display

# Start virtual display (1920x1080 resolution)
display = Display(visible=0, size=(1920, 1080))
display.start()

print("✅ Virtual display started!")
print(f"   Display: {display.display}")
```

## Step 3: Test Chrome is Working
Quick test to make sure Chrome works.

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service

# Configure Chrome
chrome_options = Options()
chrome_options.add_argument('--no-sandbox')
chrome_options.add_argument('--disable-dev-shm-usage')
chrome_options.add_argument('--disable-gpu')
chrome_options.add_argument('--window-size=1920,1080')

# Test launch
try:
    driver = webdriver.Chrome(options=chrome_options)
    driver.get('https://www.google.com')
    print(f"✅ Chrome works! Title: {driver.title}")
    driver.quit()
except Exception as e:
    print(f"❌ Chrome failed: {e}")
```

## Step 4: Upload Your Automation Code
Upload your entire automation folder as a ZIP file.

```python
from google.colab import files
import zipfile
import os

print("📁 Please upload your automation code as a ZIP file...")
print("   (Create ZIP of your 'automation tool' folder first)")
print()

# Upload file
uploaded = files.upload()

# Extract ZIP
for filename in uploaded.keys():
    if filename.endswith('.zip'):
        print(f"\n📦 Extracting {filename}...")
        with zipfile.ZipFile(filename, 'r') as zip_ref:
            zip_ref.extractall('/content/automation')
        print("✅ Extracted to /content/automation")

# List files
!ls -la /content/automation
```

## Step 6: Run Video Generation!
Enter your video idea and let it run.

```python
import os
import sys

# Change to automation directory
os.chdir('/content/automation')
sys.path.insert(0, '/content/automation')

# Get user input
video_idea = input("Enter your video idea: ")
num_scenes = int(input("Number of scenes (default 5): ") or "5")

print(f"\n🎬 Starting video generation...")
print(f"   Idea: {video_idea}")
print(f"   Scenes: {num_scenes}")
print("=" * 50)

# Run the automation
!python master_manager.py --type character --idea "{video_idea}" --scenes {num_scenes} --headless --no-proxy
```

## Step 7: Download Your Video
Download the generated video to your computer.

```python
from google.colab import files
import glob

# Find generated videos
videos = glob.glob('/content/automation/output/**/*.mp4', recursive=True)

if videos:
    print("📹 Found videos:")
    for i, v in enumerate(videos):
        print(f"   {i+1}. {v}")
    
    # Download the latest one
    latest = max(videos, key=os.path.getctime)
    print(f"\n⬇️ Downloading: {latest}")
    files.download(latest)
else:
    print("❌ No videos found. Check the output directory:")
    !ls -la /content/automation/output/
```
